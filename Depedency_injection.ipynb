```python
#depedency injection and depedency inversion

class EmailService:
  def send(self,message):
    print(f"your message is here :{message}")

class OrderService:
  def __init__(self) -> None:
    self.nofitication_service=EmailService()

  def orderPlaced(self):
    print("Order placed successfully!")
    self.nofitication_service.send("Your order has been placed.")

order_service=OrderService()
order_service.orderPlaced()
```

```python
class EmailService:
  def send(self,message):
    print(f"Email sent : {message}")

class SMSService:
  def send(self,message):
    print(f"SMS sent : {message}")


class OrderService:
  def __init__(self,nofitication_service) -> None:
    self.nofiticationService=nofitication_service

  def orderPlaced(self):
    print("order placed successfully")
    self.nofiticationService.send("your order has been placed")

emailService=EmailService()
smsService=SMSService()

orderService=OrderService(smsService)
orderService.orderPlaced()
```

```python
from abc import ABC,abstractmethod

class NotificationService(ABC):
  @abstractmethod
  def send(self,message):
    pass

class PushNotificationService(NotificationService):
  def send(self,message):
    print("Push notification sent: Your order has been placed.")

class EmailService(NotificationService):
  def send(self,message):
    print(f"Email : {message}")

class SMSService(NotificationService):
  def send(self,message):
    print(f"SMS : {message}")


class OrderService:
  def __init__(self,notification_service:NotificationService):
    self.notification_service=notification_service

  def orderPlaced(self):
    print("order placed")
    self.notification_service.send("order has beed created")


#for email

emailService=EmailService()
orderservice=OrderService(emailService)
orderservice.orderPlaced()

#for sms

smsService=SMSService()
orderservice=OrderService(smsService)
orderservice.orderPlaced()


push_service = PushNotificationService()
order_service = OrderService(push_service)
order_service.orderPlaced()
```
